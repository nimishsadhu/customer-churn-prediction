# Customer Churn Prediction - Data Inspection
## Objective

Before building a machine learning model, it is important to understand the data we are working with.

In this notebook, I will inspect the Telco Customer Churn dataset to understand its structure, data types, data quality, target distribution, and the meaning of different features.

The main goals of this inspection are to:

- Understand what each column represents
- Identify numerical and categorical features
- Check for missing and duplicate values
- Find columns with incorrect or unexpected data types
- Understand the distribution of the target variable
- Identify irrelevant features and potential data leakage
- Detect any data-quality issues that need to be addressed before preprocessing

The findings from this notebook will be used to make informed decisions during the preprocessing and modeling stages.

## 7. Duplicate Records and Unique Identifiers

Duplicate records can affect our analysis and model training. I will also check whether `CustomerID` uniquely identifies each customer.

## 8. Categorical Feature Inspection

Categorical features can contain different categories, unexpected values, or inconsistent representations. I will inspect their unique values to understand what each feature actually contains.

```python
# Categorical columns
categorical_cols = df.select_dtypes(include = ['object' ,'category'])
```

```python
# Checking unique values of categorical columns
for col in categorical_cols.columns:
    print(f"\n{col}")
    print(df[col].unique())
```

```python
# Checking how many customers stayed and how many churned
df["Churn Value"].value_counts()
```
